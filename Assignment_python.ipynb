```python
# question 7.3
def print_abs(val):
    if val < 0:
        print(-val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)
```

```python
# question 7.4
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)

# Why doesn't x change?

# When inc_val(x) is called, val gets the value 7. Inside the function, val = val + 1 makes val refer to a new integer object (8), but it does not change x.

# So after the function finishes, x is still 7.

# This is because integers are immutable in Python, and reassigning a function parameter does not reassign the original variable.
```

```python
# question 7.5
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function:", val1, val2)

x = 6
y = 3

swap_message(x, y)

print("Original variables:", x, y)

# Why?

# val1 and val2 are local variables inside the function. Swapping them changes only what they refer to inside the function.

# The original x and y remain unchanged: x = 6 and y = 3.
```

Topic 8: Scope — Local vs Global Variables

```python
# question 8.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(val)

# Why?

# val is a local variable of my_abs(). It exists only inside the function. When we write print(val) outside the function, Python doesn't know what val refers to, so it raises a NameError.
```

```python
# If you want to print the result, you need to call the function

print(my_abs(-5))
```

```python
# question 8.2
counter = 0

def show_counter():
    print(counter)

show_counter()

# Why?

# counter is a global variable because it is created outside the function. The function can read the global variable directly, so show_counter() successfully prints 0
```

```python
# question 8.3
counter = 0

def increment_counter():
    counter = counter + 1

increment_counter()

# Why?

# When Python sees counter = counter + 1 inside the function, it treats counter as a local variable. But you're trying to use counter before giving that local variable a value, so Python raises UnboundLocalError
```

```python
# To modify the global variable, you would need the global keyword
def increment_counter():
    global counter
    counter = counter + 1
```

```python
# question 8.4
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)

# Why?

# global counter tells Python that the counter inside the function refers to the global variable, so the function can modify it.
```
